# **Predicting High-Cost Medicare Beneficiaries: A Machine Learning Comparison Using CMS Synthetic Claims Data**

Medicare covers 65M+ beneficiaries and exceeds $100B in annual Part D spending, with a small subset of patients driving a disproportionate share of total costs. Early identification of these high-cost beneficiaries enables targeted care management, supports value-based care contracts, and informs specialty drug program design.

This project uses CMS De-SynPUF Sample 9, a 0.25% subsample containing ~115k beneficiaries per year (2008–2010) and millions of associated claims, including 4.7M Carrier, 66k inpatient, 791k outpatient, and 5.55M PDE records. These files span inpatient, outpatient, professional, and pharmacy utilization to support comprehensive feature development.

Although the dataset is synthetic and based on ICD-9–era patterns, its structure closely mirrors real CMS claims, making it useful for methodological testing. This project will compare different ML algorithm performances on high-cost beneficiaries predictions and evaluate such outputs through AUC, calibration, and other relevant metrics.

```python
outpatient = outpatient.rename(columns={
    'avg_claim_duration': "avg_op_claim_duration",
    'max_claim_duration': "max_op_claim_duration",
    'total_claim_duration': "total_op_claim_duration",
    'max_chronic_code_counts': "max_op_chronic_code_counts"
})
```

```python
merged_claim = reconcile_disease_flags([carrier, inpatient, outpatient])
merged_claim.head()
```

```python
pde = pde.rename(columns={
    "SRVC_year": "claim_year"
})
```

```python
merged_claim2 = merged_claim.merge(pde, left_on=["DESYNPUF_ID", "claim_year"], right_on=["DESYNPUF_ID", "claim_year"], how="outer")
```

```python
merged_claim2["has_carrier_record"] = merged_claim2["num_carrier_claims"].notna()
merged_claim2["has_ip_record"] = merged_claim2["num_ip_claims"].notna()
merged_claim2["has_op_record"] = merged_claim2["num_op_claims"].notna()
merged_claim2["has_pde_record"] = merged_claim2["num_pde_claims"].notna()
merged_claim2[["has_carrier_record", "has_ip_record", "has_op_record", "has_pde_record"]].value_counts()
```

```python
merged_claim2["num_service_types"] = merged_claim2[["has_carrier_record", "has_ip_record", "has_op_record", "has_pde_record"]].sum(axis=1)
merged_claim2["num_service_types"].value_counts()
```

```python
merged_claim2 = merged_claim2.rename(columns={
    "claim_year": "year"})
```

```python
sp_mapping = {
    "SP_ALZHDMTA": "has_Alzheimers/Dementia",
    "SP_CHF": "has_Heart Failure",
    "SP_CHRNKIDN": "has_CKD",
    "SP_CNCR": "has_Cancer",
    "SP_COPD": "has_COPD",
    "SP_DEPRESSN": "has_Depression",
    "SP_DIABETES": "has_Diabetes",
    "SP_ISCHMCHT": "has_Ischemic Heart Disease",
    "SP_OSTEOPRS": "has_Osteoporosis",
    "SP_RA_OA": "has_Rheumatoid Arthritis/Osteoarthritis",
    "SP_STRKETIA": "has_Stroke/TIA"
}
```

```python
sp_cols = [
    "SP_ALZHDMTA",
    "SP_CHF",
    "SP_CHRNKIDN",
    "SP_CNCR",
    "SP_COPD",
    "SP_DEPRESSN",
    "SP_DIABETES",
    "SP_ISCHMCHT",
    "SP_OSTEOPRS",
    "SP_RA_OA",
    "SP_STRKETIA"
]
for col in sp_cols:
    print(beneficiaries[col].value_counts(dropna=False))
```

```python
beneficiaries["SP_COPD"] = beneficiaries["SP_COPD"].replace({1:1, 2:0})
```

```python
patient_year_data = beneficiaries.merge(merged_claim2, left_on=["DESYNPUF_ID", "year"], right_on=["DESYNPUF_ID", "year"], how="left")
patient_year_data.head()
```

```python
for sp_col, disease_col in sp_mapping.items():

    if disease_col in patient_year_data.columns:

        patient_year_data[disease_col] = (
            patient_year_data[[disease_col, sp_col]]
            .fillna(0)
            .max(axis=1)
            .astype(int)
        )
```
